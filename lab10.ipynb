# Короткі теоретичні відомості

Кодування Гафмена — це алгоритм стиснення даних без втрат.

Його ідея полягає в тому, що символам з більшою частотою призначаються коротші коди, а символам з меншою частотою — довші.

Коди Гафмена є префіксними, тобто жоден код не є префіксом іншого.

# Побудова дерева Гафмена

```python
import heapq
```

```python
class node:

    def __init__(self, freq, symbol, left=None, right=None):

        self.freq = freq
        self.symbol = symbol
        self.left = left
        self.right = right
        self.huff = ''

    def __lt__(self, nxt):

        return self.freq < nxt.freq
```

```python
def printNodes(node, val=''):

    newVal = val + str(node.huff)

    if node.left:
        printNodes(node.left, newVal)

    if node.right:
        printNodes(node.right, newVal)

    if not node.left and not node.right:
        print(f"{node.symbol} -> {newVal}")
```

```python
chars = ['a', 'b', 'c', 'd', 'e', 'f']

freq = [5, 9, 12, 13, 16, 45]
```

```python
nodes = []

for x in range(len(chars)):

    heapq.heappush(nodes, node(freq[x], chars[x]))
```

```python
while len(nodes) > 1:

    left = heapq.heappop(nodes)
    right = heapq.heappop(nodes)

    left.huff = 0
    right.huff = 1

    newNode = node(left.freq + right.freq,
                   left.symbol + right.symbol,
                   left,
                   right)

    heapq.heappush(nodes, newNode)
```

```python
printNodes(nodes[0])
```

# Часова складність алгоритму

Часова складність алгоритму Гафмена:

`O(n log n)`

де `n` — кількість унікальних символів.

# Просторова складність

Просторова складність:

`O(n)`

### 1. Що таке жадібні алгоритми?

Жадібні алгоритми — це алгоритми, які на кожному кроці обирають локально оптимальне рішення.

### 2. Що таке префіксний код?

Префіксний код — це код, у якому жодне кодове слово не є префіксом іншого.

У коді Гафмена використовується двійковий префіксний код.

### 3. Як пов’язана купа з чергою з пріоритетами?

Купа використовується для реалізації черги з пріоритетами.

Вона дозволяє швидко отримувати елемент з найменшим пріоритетом.

### 4. Що таке стиснення даних?

Стиснення даних — це процес зменшення розміру даних для економії пам’яті та пришвидшення передачі інформації.
