Дослідження Bubble Sort та Insertion Sort
1. Теорія
Bubble Sort

Bubble Sort порівнює сусідні елементи та міняє їх місцями, якщо вони стоять у неправильному порядку.

Приклад:
[5,3,1]→[3,5,1]→[3,1,5]→[1,3,5]

Асимптотика Bubble Sort
найкращий випадок:

O(n)

середній випадок:

O(n2)

найгірший випадок:

O(n2)

Квадратична складність виникає через два вкладені цикли.
Insertion Sort вставляє кожен елемент у правильне місце у вже відсортованій частині масиву.

Приклад:

[5,3,1]

Після вставки 3:

[3,5,1]

Після вставки 1:

[1,3,5]
Асимптотика Insertion Sort
найкращий випадок:
O(n)
середній випадок:
O(n2)
найгірший випадок:
O(n2)

Чому саме O(n2)
Алгоритм містить два вкладені цикли:

```python
import random
import time
import matplotlib.pyplot as plt
```

3. Реалізація Bubble Sort

```python
def bubble_sort(arr):

    a = arr.copy()
    n = len(a)

    for i in range(n):

        for j in range(0, n - i - 1):

            if a[j] > a[j + 1]:

                a[j], a[j + 1] = a[j + 1], a[j]

    return a
```

4. Реалізація Insertion Sort

```python
def insertion_sort(arr):

    a = arr.copy()

    for i in range(1, len(a)):

        key = a[i]
        j = i - 1

        while j >= 0 and a[j] > key:

            a[j + 1] = a[j]
            j -= 1

        a[j + 1] = key

    return a
```

5. Розміри масивів

```python
sizes = [5, 10, 50, 100, 500,
         1000, 2000, 3000,
         4000, 5000]
```

6. Вимірювання часу

```python
insert_times = []
bubble_times = []

for n in sizes:

    arr = [random.randint(0, 100000)
           for _ in range(n)]

    start = time.perf_counter()

    insertion_sort(arr)

    end = time.perf_counter()

    insert_times.append(end - start)

    start = time.perf_counter()

    bubble_sort(arr)

    end = time.perf_counter()

    bubble_times.append(end - start)

    print(f"n = {n}")
    print(f"Insertion: {insert_times[-1]:.6f}")
    print(f"Bubble: {bubble_times[-1]:.6f}")
```

7. Побудова графіків

```python
plt.figure(figsize=(10, 6))

plt.plot(
    sizes,
    insert_times,
    marker='o',
    label='Tinsert(n)'
)

plt.plot(
    sizes,
    bubble_times,
    marker='s',
    label='Tbubble(n)'
)

plt.xlabel("Розмір масиву")

plt.ylabel("Час (сек)")

plt.title("Порівняння сортувань")

plt.grid(True)

plt.legend()

plt.show()
```

8. Висновок
Обидва алгоритми мають складність:
O(n2)
При збільшенні розміру масиву час роботи швидко зростає.
Bubble Sort працює повільніше за Insertion Sort.
