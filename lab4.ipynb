## Теоретичні відомості
Рекурсія — це спосіб організації алгоритму, при якому функція викликає сама себе.
Стратегія «розділяй і володарюй» полягає у:
1. поділі задачі на менші підзадачі;
2. рекурсивному розв’язанні підзадач;
3. об’єднанні результатів.

```python
def factorial(n):
    
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)
    
print(factorial(5))
```

Функція factorial() викликає сама себе, поки не досягне базового випадку.

Асимптотика алгоритму:
\[
O(n)
\]
оскільки виконується n рекурсивних викликів.

```python
def merge_sort(arr):

    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    result = []

    i = j = 0

    while i < len(left) and j < len(right):

        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


arr = [5, 2, 8, 1, 3]

print(merge_sort(arr))
```

Merge Sort використовує стратегію «розділяй і володарюй».

Масив ділиться на дві частини, які рекурсивно сортуються, після чого об’єднуються.

Рекурентне співвідношення:
\[
T(n)=2T\left(\frac{n}{2}\right)+O(n)
\]

За головною теоремою рекурсії:
\[
O(n \log n)
\]
