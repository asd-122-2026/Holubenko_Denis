## Теоретичні відомості
Пошук — це процес знаходження елемента в масиві або іншій структурі даних.
У лабораторній роботі розглядаються:
- лінійний пошук;
- бінарний пошук.

## Лінійний пошук
Лінійний пошук послідовно перевіряє всі елементи масиву, поки не знайде потрібний.
Асимптотична складність:
O(n)

```python
def linear_search(arr, target):

    for i in range(len(arr)):

        if arr[i] == target:
            return i

    return -1


arr = [3, 7, 2, 9, 5]

print(linear_search(arr, 9))
```

## Бінарний пошук
Бінарний пошук працює лише для відсортованого масиву.

Алгоритм:
1. знаходить середній елемент;
2. порівнює його з шуканим;
3. відкидає половину масиву.

Асимптотична складність:
O(log n)

```python
def binary_search(arr, target):

    left = 0
    right = len(arr) - 1

    while left <= right:

        mid = (left + right) // 2

        if arr[mid] == target:
            return mid

        elif arr[mid] < target:
            left = mid + 1

        else:
            right = mid - 1

    return -1

arr = [1, 2, 3, 5, 7, 9]

print(binary_search(arr, 7))
```

## Дослідження часу виконання

```python
import random
import time
import pandas as pd
```

```python
def linear_search_test(sizes):

    result = []

    for n in sizes:

        arr = list(range(n))

        target = random.randint(0, n - 1)

        start = time.perf_counter()

        linear_search(arr, target)

        end = time.perf_counter()

        result.append([n, end - start])

    return pd.DataFrame(result, columns=["n", "time"])
```

```python
def binary_search_test(sizes):

    result = []

    for n in sizes:

        arr = list(range(n))

        target = random.randint(0, n - 1)

        start = time.perf_counter()

        binary_search(arr, target)

        end = time.perf_counter()

        result.append([n, end - start])

    return pd.DataFrame(result, columns=["n", "time"])
```

```python
sizes = [10, 100, 1000, 5000, 10000, 50000]

linear_df = linear_search_test(sizes)

binary_df = binary_search_test(sizes)

print(linear_df)

print(binary_df)
```

## Побудова графіків

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.plot(
    linear_df["n"],
    linear_df["time"],
    marker='o',
    label="Linear Search"
)

plt.plot(
    binary_df["n"],
    binary_df["time"],
    marker='s',
    label="Binary Search"
)

plt.xlabel("Розмір масиву")

plt.ylabel("Час виконання")

plt.title("Порівняння алгоритмів пошуку")

plt.grid(True)

plt.legend()

plt.show()
```

## Висновок
Лінійний пошук має складність O(n), тому час його роботи швидко зростає при збільшенні розміру масиву.

Бінарний пошук має складність O(log n), тому працює значно швидше на великих масивах.

Найефективнішим є бінарний пошук, але він потребує попереднього сортування масиву.

- висока швидкість;
- складність O(log n).
