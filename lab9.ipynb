# Наївний пошук шаблонів

Наївний алгоритм перевіряє збіг шаблону з текстом на кожній позиції.

```python
def naive_match(p, t):

    assert len(p) <= len(t)

    occurrences = []

    for i in range(0, len(t) - len(p) + 1):

        match = True

        for j in range(0, len(p)):

            if t[i + j] != p[j]:

                match = False
                break

        if match:
            occurrences.append(i)

    return occurrences
```

```python
print(naive_match('needle', 'needleneedleneedle'))
```

## Асимптотична складність

Складність наївного алгоритму:

`O(|t| * |p|)`

# Z-функція

Z-функція — це масив, де кожен елемент показує довжину найбільшого підрядка, який збігається з префіксом рядка.

```python
def z_func(s):

    Z = [len(s)] + [0] * len(s)

    assert len(s) > 1

    for i in range(1, len(s)):

        if s[i] == s[i - 1]:
            Z[1] += 1
        else:
            break

    r, l = 0, 0

    if Z[1] > 0:
        r, l = Z[1], 1

    for k in range(2, len(s)):

        assert Z[k] == 0

        if k > r:

            for i in range(k, len(s)):

                if s[i] == s[i - k]:
                    Z[k] += 1
                else:
                    break

            r, l = k + Z[k] - 1, k

        else:

            nbeta = r - k + 1
            Zkp = Z[k - l]

            if nbeta > Zkp:

                Z[k] = Zkp

            else:

                nmatch = 0

                for i in range(r + 1, len(s)):

                    if s[i] == s[i - k]:
                        nmatch += 1
                    else:
                        break

                l, r = k, r + nmatch
                Z[k] = r - k + 1

    return Z
```

```python
print(z_func('abracadabra'))
```

# Пошук підрядка за допомогою Z-функції

```python
def zMatch(p, t):

    s = p + "$" + t

    Z = z_func(s)

    occurrences = []

    for i in range(len(p) + 1, len(s)):

        if Z[i] == len(p):

            occurrences.append(i - (len(p) + 1))

    return occurrences
```

```python
t = "абабагаламага"
p = "аб"

print(zMatch(p, t))
```

## Асимптотична складність

`O(len(p) + len(t))`

# Стиснення рядка

def compress_with_z(s):

    z_vec = z_func(s)

    for i in range(1, len(s)):

        if (i + z_vec[i] == len(s)) and (z_vec[i] % i == 0):

            return s[:i]

    return s

```python
s = "абырвалгабырвалгабырвалг"

print(z_func(s))
print(compress_with_z(s))
```

# Контрольні питання
### 1. Що таке префікс-функція?

Префікс-функція — це масив, який показує довжину найбільшого власного префікса рядка, що одночасно є суфіксом.

### 2. Що таке Z-функція?

Z-функція — це масив, у якому кожен елемент показує довжину збігу підрядка з префіксом рядка.

### 3. Які існують підходи до задачі найдовшого спільного підрядка?

- динамічне програмування;
- суфіксні масиви;
- суфіксні дерева;
- хешування рядків.
### 4. Де застосовуються алгоритми на рядках?

Алгоритми на рядках застосовуються у:
- пошукових системах;
- аналізі текстів;
- обробці природної мови;
- біоінформатиці.
